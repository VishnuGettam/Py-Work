# MultiDimensional Arrays


```python
import numpy as np
```

## 0-D Array

```python
d_array = np.array(67)

print(f"Dimension of array -  {d_array.ndim}")
print(f"Shape of the array (columns)-  {d_array.shape}")
print(f"Size of the array - {d_array.size}")
```

## 1-D Array         


A 1D array is just a list of values.

```python
d_array = np.array([2,3,5]) 

print(f"Dimension of array -  {d_array.ndim}")
print(f"Shape of the array (rows,columns) -  {d_array.shape}")
print(f"Size of the array - {d_array.size}")
```

## 2-D Array



A 2D array is a collection of rows.

```python
d_array = np.array([
    
                     [3,4,5],
                     [7,5,6] 

                    ])

print(f"Dimension of array - {d_array.ndim}")
print(f"Shape of the array (rows,columns) -  {d_array.shape}")
print(f"Size of the array - {d_array.size}")

```

## 3-D Array

  

A 3D array is simply multiple 2D arrays stacked together.




Below array means:

First dimension (Depth/Layers) = 2
Think of this as 2 tables stacked one behind the other.
Second dimension (Rows) = 2
Each table has 2 rows.
Third dimension (Columns) = 3
Each row has 3 columns.


"The layers pointing at the depth are nothing but the number of layers, which will show as a mental model of the number of tables."
"Each table will have rows and columns."

* 3D NumPy array, every table (layer) must have the same number of rows and columns *



```python
d_array = np.array([
            [
                [2,1,3],
                [5,6,1] 
            ],
            [
                [5,6,7],
                [5,1,2]
            ]
        ])

print(f"Dimension of array - {d_array.ndim}")
print(f"Shape of the array (rows,columns) -  {d_array.shape}")
print(f"Size of the array - {d_array.size}")
```

```python
user_data = np.array([

                    [   
                        ['user_name','age','location'],
                        ['vishnu','34','Ctr'],
                        ['vihaan','19','Bng'],
                        ['divya','34','Bng']
                    ]

                    ])


print(f"Dimension of array - {user_data.ndim}")
print(f"Shape of the array (rows,columns) -  {user_data.shape}")
print(f"Size of the array - {user_data.size}")
print(f"Type of the array elements - {user_data.dtype}")
print(f"Type of the array - {type(user_data)}")
print(f"Len of the array - {len(user_data)}")

```
