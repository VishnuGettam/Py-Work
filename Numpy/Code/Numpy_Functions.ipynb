## Copy vs View

| Feature                 | `view()` | `copy()`          |
| ----------------------- | -------- | ----------------- |
| New array object        | ✅ Yes    | ✅ Yes             |
| New memory for data     | ❌ No     | ✅ Yes             |
| Shares underlying data  | ✅ Yes    | ❌ No              |
| Changes affect original | ✅ Yes    | ❌ No              |
| Memory efficient        | ✅ Yes    | ❌ Less efficient  |
| Faster                  | ✅ Yes    | ❌ Slightly slower |


```python
    x1 = np.arange(1,10)
    x2 = x1.view()

    print(f"x1 - {x1}")
    print(f"x2 - {x2}\n")

    print(f"Object Identity Id(x1) - {id(x1)}")    
    print(f"Object Identity Id(x2) - {id(x2)}")

    print(f"\nIs x1 and x2 shares same memory - {np.shares_memory(x1,x2)}")

    print('----------------*******-------------------\n')

    y1 = np.arange(1,10)
    y2 = y1.copy()

    print(f"y1 - {y1}")
    print(f"y2 - {y2}\n")

    print(f"Object Identity Id(y1) - {id(y1)}")    
    print(f"Object Identity Id(y2) - {id(y2)}")

    print(f"\nIs y1 and y2 shares same memory - {np.shares_memory(y1,y2)}")
```

### sum

    matrix = np.array([     [1, 2, 3],
                            [4, 5, 6]])

    When you pass an axis to np.sum(), NumPy compresses that specific dimension by adding its elements together.
    
    axis=None: Flattens the entire array and sums every single number.
            Operation: 1 + 2 + 3 + 4 + 5 + 6
            Result: 21


    axis=0: Collapses the rows (dimension 0). The rows squish down, leaving you with the sum of each column.
            

    axis=1: Collapses the columns (dimension 1). The columns squish sideways, leaving you with the sum of each row.
            
            

```python
print(f"user_values - {user_values}")

y= np.sum(user_values,axis=None)

print(f"sum axis=Nome - {y}")
```

```python
# print sum in array as column wise

print(f"user_values - {user_values}")

y = np.sum(user_values,axis=0)

print(f" column sum -  {y}")
```

```python
# print sum in array as row wise

print(f"user_values - {user_values}")

y = np.sum(user_values,axis=1)

print(f" row  sum -  {y}")
```

## Statistics 

                      


| Statistical Measure | NumPy Function                    | Example         |
| ------------------- | --------------------------------- | --------------- |
| Mean                | `np.mean()`                       | `np.mean(a)`    |
| Median              | `np.median()`                     | `np.median(a)`  |
| Mode                | `scipy.stats.mode()`              | `stats.mode(a)` |
| Maximum             | `np.max()`                        | `np.max(a)`     |
| Minimum             | `np.min()`                        | `np.min(a)`     |
| Range               | `np.ptp()` or `np.max()-np.min()` | `np.ptp(a)`     |
| Variance            | `np.var()`                        | `np.var(a)`     |
| Standard Deviation  | `np.std()`                        | `np.std(a)`     |


```python
user_data = np.random.randint(1,9,(9))

print(f" User Data - {user_data}")

mean_value = np.mean(user_data)

print(f"Mean of User Data - {mean_value}")

```

```python
print(f" User Data - {user_data}")

median_Value = np.median(user_data)


print(f" User Data Sorted - {np.sort(user_data)}")

print(f"Median of User Data - {median_Value}")
```


NumPy itself does not provide a mode() function.

The recommended approach is to use SciPy.

```python
from scipy import stats


print(f" User Data - {user_data}")

mode_Value = stats.mode(user_data)

print(f"Mode of User Data - {mode_Value}")
```

## Standard Deviation 

```python
print(f" User Data - {user_data}")

result_std =  np.std(user_data)

print(f"Standard Deviation - {result_std}")
```

## Searching and Conditional

| **Category**            | **Function**   | **Example**       | **Purpose**                                                                      |
| ----------------------- | -------------- | ----------------- | -------------------------------------------------------------------------------- |
| **Sorting & Searching** | `np.sort()`    | `np.sort(a)`      | Sort elements                                                                    |
|                         | `np.argsort()` | `np.argsort(a)`   | Return indices that would sort the array                                         |
|                         | `np.where()`   | `np.where(a > 5)` | Return indices of elements satisfying a condition or choose values conditionally |
|                         | `np.unique()`  | `np.unique(a)`    | Return unique elements                                                           |


```python
user_data = np.arange(1,15)
print(f" User Data - {user_data}")

even_numbers = np.where(user_data % 2 ==0)

print(f" Even Numbers Indices - {even_numbers}") # print the indices where the condition is true 

print(f" Even Numbers Data - { user_data[even_numbers]}")
```
